```python
#define input molecule (For our test we use just the H2 molecule)
from pyscf import gto
mol = gto.M(
    atom ='H 0 0 0; H 0 0 0.74',
    basis ='sto-3g'
)

#Compute AO integrals

#overlap matrix
S = mol.intor('int1e_ovlp')

#one electron integrals
T = mol.intor('int1e_kin')
V = mol.intor('int1e_nuc')

h_ao = T + V

#two electron integrals

eri_ao = mol.intor('int2e')
print(eri_ao)

```

```python
# Hartree-Fock
from pyscf import scf

mf = scf.RHF(mol)
mf.kernel()

#MO coefficients
C = mf.mo_coeff

#Orbital Energies
eps = mf.mo_energy

#HF energy
hf_energy = mf.e_tot

```

```python
#Transform integrals into MO basis
import numpy as np

#one electron transformation
h_mo = C.T @ h_ao @ C

#Two electron transformation
from pyscf import ao2mo

eri_mo = ao2mo.kernel(
    mol,
    C
)
print(eri_mo)
```
